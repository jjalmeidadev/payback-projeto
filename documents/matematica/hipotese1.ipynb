## A) afirmação e hipóteses

afirmação: a variável valor do boleto não segue uma distribuição normal

Hipótese nula (H0): A variável segue a distribuição normal
Hipótese alternativa (Ha): A variável **não** segue a distribuição normal

## B) Nível de significância e relação com o valor de p

### Alfa

$\alpha$ = 5%

### relação com valor de p

p $\leq \alpha \rightarrow$ rejeitamos H₀ (a variável não é normal).
p > $\alpha \rightarrow$ não é possível rejeitar H₀ (a variável pode ser normal).


```python
# Importar bibliotecas necessárias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import scipy.stats as stats
from statsmodels.stats.diagnostic import lilliefors
```

```python
# Carregar os arquivos CSV
df_gf = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GF.csv", sep='\t')
df_gl = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GL.csv", sep='\t')
df_gm = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GM.csv", sep='\t')
df_gp = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GP.csv", sep='\t')
df_gt = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GT.csv", sep='\t')
df_gv = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GV.csv", sep='\t')
df_gw = pd.read_csv("../../data/Grupo com registro entre 07-2024 a 06-2025- GW.csv", sep='\t')
```

```python
df_gw.head() # mostra as primeiras linhas e os nomes de colunas
```

```python
# Configuração
alpha = 0.05  # nível de significância
variaveis = ['vl_boleto', 'vl_pagto', 'juros']  # ajuste para as colunas do seu dataset
```

```python
# --- Loop para rodar o teste em cada variável ---
resultados = []

for var in variaveis:
    # Garante que a coluna é numérica e remove valores ausentes
    x = pd.to_numeric(df_gw[var], errors='coerce').dropna()
    
    # Teste de Lilliefors (H0: distribuição é normal com μ e σ estimados)
    stat, p = lilliefors(x, dist='norm')
    
    # Conclusão com base no p-valor
    if p > alpha:
        conclusao = "Não rejeita H₀ (≈ normal)"
    else:
        conclusao = "Rejeita H₀ (não normal)"
    
    resultados.append([var, p, conclusao])
```

```python
# --- Cria tabela organizada ---
tabela = pd.DataFrame(resultados, columns=['Variável', 'p-valor', 'Conclusão'])
print(tabela)
```
